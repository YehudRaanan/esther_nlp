### Reconstructing the Megillat Esther Pipeline and building Zeresh

This notebook walks through the initial parsing of the Megillat Esther consonantal corpus. 
We parse the 167 verses separated into 10 chapters and preview our NLP pipeline flowchart.

```python
# Import data loaders from our modular package
from esther_nlp.data import load_chapters_and_verses, get_verse_location_map
from esther_nlp.plots import fig_01_pipeline
from pathlib import Path
from IPython.display import Image

# Load chapters and verses flat list
chapters, all_verses = load_chapters_and_verses()
print(f"Loaded {len(chapters)} chapters and {len(all_verses)} flat verses.")

# Get location coordinates map
loc_map = get_verse_location_map(chapters)
print(f"First verse coordinates: {loc_map[0]}")
print(f"Last verse coordinates: {loc_map[166]}")

# Generate the pipeline flowchart
fig_path = Path("fig_01_pipeline.png")
fig_01_pipeline(fig_path)
Image(filename=fig_path)
```
