# Chapter 6 — The Cast Shifts
### River Mentions Streamgraph

This notebook calculates protagonist name mention counts across all 10 chapters and plots a centered streamgraph river of the cast.

```python
from esther_nlp.data import load_chapters_and_verses
from esther_nlp.tokenizer import EstherBpeTokenizer
from esther_nlp.plots import fig_06_streamgraph
from pathlib import Path
from IPython.display import Image
import numpy as None
import numpy as np

chapters, all_verses = load_chapters_and_verses()
tokenizer = EstherBpeTokenizer(vocab_size=1000)
tokenizer.train(all_verses)
tokenized_by_ch = [[tokenizer.encode_to_tokens(v) for v in ch] for ch in chapters]

char_groups = {
    "המלך":     ["המלך"],
    "אחשורוש":  ["אחשורוש", "אחשורש"],
    "ושתי":     ["ושתי"],
    "אסתר":     ["אסתר", "לאסתר"],
    "מרדכי":    ["מרדכי", "ומרדכי", "למרדכי", "במרדכי"],
    "המן":      ["המן", "והמן", "להמן", "מהמן"],
    "היהודים":  ["היהודים", "ביהודים", "ליהודים"],
}
density = np.zeros((len(char_groups), 10), dtype=float)
char_names = list(char_groups.keys())

for ci, ch_tokens in enumerate(tokenized_by_ch):
    flat_tokens = [token for verse in ch_tokens for token in verse]
    for ri, (_, group) in enumerate(char_groups.items()):
        density[ri, ci] = sum(flat_tokens.count(g) for g in group)

# Generate protagonist streamgraph
fig_path = Path("fig_06_streamgraph.png")
fig_06_streamgraph(density, char_names, fig_path)
Image(filename=fig_path)
```

## Cast Density Analysis & Conclusion

### Core Observations
1. **Vashti's quick exit**: Vashti has 7 mentions in chapter 1, 3 in chapter 2, and then completely vanishes. 
2. **Esther's Chapter 6 absence**: Esther is the grand heroine, yet in the turning point of the book (Chapter 6 — the sleepless night of the king), she has **exactly one mention**. The pivot occurs in her absolute absence.
3. **Collective takeover**: `היהודים` (the Jews) have virtually no mentions through the first half, but swell to 12 mentions in chapter 8 and 21 mentions in chapter 9.

### Narrative and Literary Conclusion
This tracking reveals the macro-narrative shift of the book. The Megillah starts as a story about specific royal figures (Vashti, Achashverosh) and court politics, but resolves into a collective story about a people. The heroine initiates the opportunity, but by chapter 9, the protagonist is the collective `היהודים` (the Jews). The streamgraph perfectly charts this narrative expansion.
